```python
import asyncio, httpx, uvloop
host = 'http://192.168.50.128:11434'
headers = {"Content-Type": "application/json"}

client = httpx.AsyncClient(
            base_url = host,
            headers = headers,
            timeout = httpx.Timeout(connect=10.0, read=600.0, write=60.0,
                                  pool=10.0)
        )
a = await client.get("/api/tags")
a.raise_for_status()
```

```python
c = httpx.AsyncClient(base_url="http://192.168.50.128:11434", timeout=10)
try:
    r = await c.get("/api/tags")
    print("uvloop+httpx: OK", r.status_code)
except Exception as e:
    print("uvloop+httpx: failed:", type(e).__name__, repr(e))
finally:
    await c.aclose()
```
